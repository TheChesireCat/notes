```python
#| label: fig-companion-ws
import matplotlib.pyplot as plt
import networkx as nx
import numpy as np

ps = np.logspace(-4, 0, 25)
C0 = nx.average_clustering(nx.watts_strogatz_graph(1000, 10, 0, seed=0))
L0 = nx.average_shortest_path_length(nx.watts_strogatz_graph(1000, 10, 0, seed=0))
C, L = [], []
for p in ps:
    G = nx.connected_watts_strogatz_graph(1000, 10, p, seed=1)
    C.append(nx.average_clustering(G) / C0)
    L.append(nx.average_shortest_path_length(G) / L0)
fig, ax = plt.subplots(figsize=(5, 3.2))
ax.semilogx(ps, C, 'o', label='$C(p)/C(0)$')
ax.semilogx(ps, L, 's', label='$L(p)/L(0)$')
ax.set_xlabel('rewiring probability $p$')
ax.legend(frameon=False)
plt.show()
```
